# Recipe Journal: data model exploration

A shared notebook for trying out sample recipe data and discussing how user stories map to data. Keep production logic in `app/`; this notebook is for exploration. Use fictional sample data only.

```python
from dataclasses import asdict, dataclass
from pprint import pprint

@dataclass
class Ingredient:
    name: str
    quantity: float
    unit: str = ''

@dataclass
class Recipe:
    id: str
    title: str
    ingredients: list[Ingredient]
    steps: list[str]
    servings: int
    tags: list[str]

sample = Recipe(
    id='tomato-pasta',
    title='Tomato Pasta',
    ingredients=[Ingredient('Tomatoes', 2, 'cups'), Ingredient('Olive oil', 1, 'tbsp')],
    steps=['Boil pasta.', 'Warm tomatoes and olive oil.', 'Combine and serve.'],
    servings=2,
    tags=['Dinner', 'Vegetarian'],
)
pprint(asdict(sample))
```

## Team questions

- Which fields are required before a recipe can be saved?
- Should ingredients be normalized into quantity, unit, and name? How will fractions and units work?
- Which data belongs to a recipe, and which belongs to a user's saved/favorite state?
- What should happen to meal plans or shopping lists if a recipe is edited or removed?
